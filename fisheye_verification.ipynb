```python
fisheye_params = {
    "width": 2592,
    "height": 1944,
    "cx": 1296.0,
    "cy": 972.0,
    "radius": 952.56,
    "theta_max_rad": 1.5708
}

cam_settings = [
    {"cx": 1296.0, "cy": 972.0, "fov": 90.0, "pitch": 58.0, "yaw": -18.0},
    {"cx": 1296.0, "cy": 972.0, "fov": 90.0, "pitch": 54.0, "yaw": 48.0},
    {"cx": 1296.0, "cy": 972.0, "fov": 90.0, "pitch": 28.0, "yaw": 150.0},
    {"cx": 1296.0, "cy": 972.0, "fov": 90.0, "pitch": 58.0, "yaw": 294.0},
]

# fisheye_path = "capture_15/fisheye_1_233.jpg"
fisheye_path = "cam_4/cam_233_fisheye.png"
BEV = cv2.imread("ohlf.png")
# h = np.load('cam_229_homo.npy')
# h = np.array([[-4.11729955e-01,  3.01608102e-01,  1.54761742e+03],
#         [-3.84689520e-01, -3.98373224e-02,  1.23992428e+03],
#         [-2.23425190e-04,  1.12290017e-04,  1.11809204e+00]])
# h = np.array([[-1.38388209e-01,  2.76194190e-01,  1.53250265e+03],
#  [-2.88215404e-01, -1.15516510e-01,  1.36087820e+03],
#  [-2.17305932e-06,  5.57611481e-05,  1.11497239e+00]]) #234
# h = np.array( [[-2.32554929e+00, -2.77936681e+00,  1.42015459e+04],
#  [-3.18907100e+00, -3.81594301e-01,  7.69143249e+03],
#  [-2.55220223e-03, -3.19491068e-03,  1.32098001e+01]])
# h = np.array([[-2.37453084e+00, -2.83791116e+00,  1.45006779e+04],
#  [-3.25624257e+00 ,-3.89632652e-01,  7.85343983e+03],
#  [-2.60595854e-03, -3.26220733e-03,  1.34880427e+01]]) #233
# h = np.array([[-4.39550251e-01, -2.41187773e-01,  2.99288453e+03],
#         [ 3.33132247e-01, -5.06169903e-01,  1.03228844e+03],
#         [-3.40592543e-05, -5.36871055e-06,  1.50711356e+00]])
#Homography h_kafka_234_to_233_to_229_bev.npy
# dh
# h = np.array([[-6.32932107e-02, -8.17181389e-02,  4.58875033e+02],
#  [-1.03317330e-01, -2.44841831e-03,  2.45706919e+02],
#  [-7.67264578e-05, -1.02846447e-04,  4.38582250e-01]]) #233
# h = np.array([[-6.05690077e-01,  8.47414334e-01,  7.87288654e+03],
#  [-1.34305545e+00, -8.33081690e-01,  6.81989115e+03],
#  [ 1.54523918e-05, -8.48904033e-05,  5.77218124e+00]]) #234
# h = np.array([[ 3.11724993e+02,  8.12285445e+02, -2.85522286e+06],
#  [ 6.50321971e+02,  1.98224735e+02, -1.92604570e+06],
#  [ 3.98254224e-01,  8.65135649e-01, -2.59925747e+03]]) #full refine 234
# h= np.array([[-4.15311101e-01 , 1.70693621e-01 , 4.99670491e+03],
#  [-8.08552754e-01, -6.20731051e-01,  4.14594219e+03],
#  [-5.79772710e-05, -2.83105932e-04,  3.69650497e+00]]) # correct sh 234
h = np.array([[-1.33738226e+01 ,-3.36040604e+01 , 1.84275435e+05],
 [-3.98040255e+01 , 9.54950726e-03 , 1.01087605e+05],
 [-2.32920743e-02, -4.46829518e-02 , 1.81982987e+02]])# correct sh 233
# h=np.array([[-1.76053994e-01, -2.10399165e-01,  1.07507769e+03],
#         [-2.41419105e-01, -2.88864727e-02,  5.82252154e+02],
#         [-1.93209717e-04, -2.41856879e-04,  1.00000000e+00]])
# h=np.array([[-7.34990253e-02, -1.84650058e-01,  1.01259845e+03],
#  [-2.18727747e-01,  5.45498552e-05,  5.55478633e+02],
#  [-1.27997107e-04, -2.45529664e-04,  1.00000000e+00]])
fisheye_img = cv2.imread(fisheye_path)


def extract_directional_crop(fisheye_img,yaw_deg,pitch_deg,fov_deg=90,out_w=1280,out_h=720,fisheye_params=fisheye_params):

    cx = fisheye_params["cx"]
    cy = fisheye_params["cy"]
    f_fish = fisheye_params["radius"] / fisheye_params["theta_max_rad"]

    yaw = np.deg2rad(yaw_deg)
    pitch = np.deg2rad(pitch_deg)
    fov = np.deg2rad(fov_deg)
    fx = (out_w / 2) / np.tan(fov / 2)
    fy = fx
    cx_o = out_w / 2
    cy_o = out_h / 2
    x = np.linspace(0, out_w - 1, out_w)
    y = np.linspace(0, out_h - 1, out_h)
    xv, yv = np.meshgrid(x, y)
    x_cam = (xv - cx_o) / fx
    y_cam = (yv - cy_o) / fy
    z_cam = np.ones_like(x_cam)

    norm = np.sqrt(x_cam**2 + y_cam**2 + z_cam**2)
    x_cam /= norm
    y_cam /= norm
    z_cam /= norm

    R_yaw = np.array([
        [np.cos(-yaw), -np.sin(-yaw), 0],
        [np.sin(-yaw),  np.cos(-yaw), 0],
        [0,            0,           1]
    ])

    R_pitch = np.array([
        [1, 0,             0],
        [0, np.cos(pitch), -np.sin(pitch)],
        [0, np.sin(pitch),  np.cos(pitch)]
    ])

    R_fix = np.array([
        [0, -1, 0],
        [1,  0, 0],
        [0,  0, 1]
    ])

    R = R_fix@R_yaw @ R_pitch

    rays = np.stack((x_cam, y_cam, z_cam), axis=-1)
    rays_rot = rays @ R.T

    theta = np.arccos(rays_rot[..., 2])
    phi = np.arctan2(rays_rot[..., 1], rays_rot[..., 0])
    r = f_fish * theta
    u = cx + r * np.cos(phi)
    v = cy + r * np.sin(phi)
    map_x = u.astype(np.float32)
    map_y = v.astype(np.float32)
    rect_img = cv2.remap(fisheye_img, map_x, map_y, cv2.INTER_LINEAR)
    return rect_img

def compute_basis(yaw_deg, pitch_deg):
    yaw_deg_corrected = yaw_deg - 90.0
    yaw, pitch = np.deg2rad([yaw_deg_corrected, pitch_deg])

    Rz = np.array([
        [np.cos(-yaw), -np.sin(-yaw), 0],
        [np.sin(-yaw),  np.cos(-yaw), 0],
        [0,             0,            1]
    ])
    Rx = np.array([
        [1, 0,           0],
        [0, np.cos(pitch), -np.sin(pitch)],
        [0, np.sin(pitch),  np.cos(pitch)]
    ])
    R_ = Rx @ Rz
    R = np.diag([1, -1, 1]) @ R_
    return R

def rect_to_fisheye_point(u, v, view_params, fisheye_params):
    Hr, Wr = rect_h, rect_w
    cx_r, cy_r = Wr / 2.0, Hr / 2.0

    fov = view_params["fov"]
    f = (Wr / 2.0) / np.tan(np.deg2rad(fov / 2.0))

    x = (u - cx_r) / f
    y = (v - cy_r) / f
    z = 1.0

    ray_rect = np.array([x, y, z], dtype=np.float64)
    ray_rect /= np.linalg.norm(ray_rect)

    Rot = compute_basis(view_params["yaw"], view_params["pitch"])
    ray_fish = Rot.T @ ray_rect
    Xf, Yf, Zf = ray_fish

    theta = np.arccos(np.clip(Zf, -1.0, 1.0))
    phi = np.arctan2(Yf, Xf)

    radius, theta_max = fisheye_params["radius"], fisheye_params["theta_max_rad"]
    cx_f, cy_f = fisheye_params["cx"], fisheye_params["cy"]

    r = (theta / theta_max) * radius
    u_f = cx_f + r * np.cos(phi)
    v_f = cy_f - r * np.sin(phi)
    return (float(u_f), float(v_f))

def fisheye_to_rect_point(u_f, v_f, view_params, fisheye_params):
    cx_f, cy_f = fisheye_params["cx"], fisheye_params["cy"]
    radius, theta_max = fisheye_params["radius"], fisheye_params["theta_max_rad"]

    dx, dy = u_f - cx_f, cy_f - v_f
    r = np.sqrt(dx ** 2 + dy ** 2)
    theta = (r / radius) * theta_max
    phi = np.arctan2(dy, dx)

    sin_t = np.sin(theta)
    ray_fish = np.array([
        sin_t * np.cos(phi),
        sin_t * np.sin(phi),
        np.cos(theta)
    ], dtype=np.float64)

    Rot = compute_basis(view_params["yaw"], view_params["pitch"])
    ray_rect = Rot @ ray_fish

    Xc, Yc, Zc = ray_rect
    if Zc <= 0:
        return None

    Hr, Wr = rect_h, rect_w
    fov = view_params["fov"]
    f = (Wr / 2.0) / np.tan(np.deg2rad(fov / 2.0))
    cx_r, cy_r = Wr / 2.0, Hr / 2.0

    u = f * (Xc / Zc) + cx_r
    v = f * (Yc / Zc) + cy_r
    if 0 <= u < Wr and 0 <= v < Hr:
        return (float(u), float(v))
    return None

views =[]
for cam in cam_settings:
    crop = extract_directional_crop(fisheye_img,cam["yaw"],cam["pitch"])
    views.append(crop)
rect_h, rect_w = views[0].shape[:2]
dis_rect_w, dis_rect_h = 2560, 2560
fov = 160.0
fov_RAD = np.deg2rad(fov)
focal = (dis_rect_w / 2.0) / np.tan(fov_RAD / 2.0)

dis_K_rect = np.array([
    [focal, 0.0, dis_rect_w / 2.0],
    [0.0, focal, dis_rect_h / 2.0],
    [0.0, 0.0, 1.0]
])

f_fish = fisheye_params["radius"] / fisheye_params["theta_max_rad"]
K_fish = np.array([
    [f_fish, 0.0, fisheye_params["cx"]],
    [0.0, f_fish, fisheye_params["cy"]],
    [0.0, 0.0, 1.0]
])
D_fish = np.zeros((4, 1)) 


scale_BEV = 0.5
scale_rect = 0.5
FISH_W_DISP, FISH_H_DISP = 1280, 720

RECT_TILE_W_DISP, RECT_TILE_H_DISP = int(rect_w * scale_rect), int(rect_h * scale_rect)

bev_display = cv2.resize(BEV, (0, 0), fx=scale_BEV, fy=scale_BEV)
views_disp = [cv2.resize(v, (0, 0), fx=scale_rect, fy=scale_rect) for v in views]
fisheye_display = cv2.resize(fisheye_img, (FISH_W_DISP, FISH_H_DISP))

scale_fx = fisheye_params["width"] / FISH_W_DISP
scale_fy = fisheye_params["height"] / FISH_H_DISP

def make_rect_grid_disp():
    grid = np.ones((RECT_TILE_H_DISP * 2, RECT_TILE_W_DISP * 2, 3), dtype=np.uint8) * 30
    grid[0:RECT_TILE_H_DISP, 0:RECT_TILE_W_DISP] = views_disp[0]
    grid[0:RECT_TILE_H_DISP, RECT_TILE_W_DISP:] = views_disp[1]
    grid[RECT_TILE_H_DISP:, 0:RECT_TILE_W_DISP] = views_disp[2]
    grid[RECT_TILE_H_DISP:, RECT_TILE_W_DISP:] = views_disp[3]
    return grid

rect_grid_disp = make_rect_grid_disp()

def draw_text(img, text, pos, color=(255, 255, 255)):
    font = cv2.FONT_HERSHEY_COMPLEX
    scale = 0.6
    thickness = 1
    x, y = int(pos[0]), int(pos[1])
    cv2.putText(img, text, (x + 1, y + 1), font, scale, (0, 0, 0), thickness + 1, cv2.LINE_AA)
    cv2.putText(img, text, (x, y), font, scale, color, thickness, cv2.LINE_AA)

rect_pts, bev_pts, fish_pts = [], [], []

def on_mouse(e, x, y, f, p):
    global rect_pts, bev_pts, fish_pts
    if e == cv2.EVENT_RBUTTONDOWN:
        rect_pts.clear(); bev_pts.clear(); fish_pts.clear()
        show(rect_grid_disp, fisheye_display, bev_display); return
    if e != cv2.EVENT_LBUTTONDOWN or x < FISH_W_DISP: return

    xr = x - FISH_W_DISP; xi, yi = xr % RECT_TILE_W_DISP, y % RECT_TILE_H_DISP
    vid = (y >= RECT_TILE_H_DISP) * 2 + (xr >= RECT_TILE_W_DISP)
    u, v = xi / scale_rect, yi / scale_rect
    rect_pts.append((vid, u, v))

    u_f, v_f = rect_to_fisheye_point(u, v, cam_settings[vid], fisheye_params)
    fish_pts.append((u_f, v_f))

    pt = np.array([[[u_f, v_f]]], np.float32)
    und = cv2.fisheye.undistortPoints(pt, K_fish, D_fish, P=dis_K_rect)
    bx, by = cv2.perspectiveTransform(und, h)[0,0]
    bev_pts.append((int(bx), int(by)))

    show_poly()

def show(rect_img, fish_img, bev_img):
    canvas = np.ones((max(FISH_H_DISP, rect_img.shape[0]),
                      FISH_W_DISP + rect_img.shape[1], 3), np.uint8)*255
    canvas[:FISH_H_DISP, :FISH_W_DISP] = fish_img
    canvas[:rect_img.shape[0], FISH_W_DISP:] = rect_img
    cv2.imshow("Unified", canvas); cv2.imshow("BEV", bev_img)

def show_poly():
    r, f, b = rect_grid_disp.copy(), fisheye_display.copy(), bev_display.copy()

    # rect polyline
    for i,(vid,u,v) in enumerate(rect_pts):
        ox, oy = (vid%2)*RECT_TILE_W_DISP, (vid//2)*RECT_TILE_H_DISP
        p=(ox+int(u*scale_rect), oy+int(v*scale_rect))
        cv2.circle(r,p,4,(0,255,0),-1)
        if i: 
            pv=rect_pts[i-1]; ox2,oy2=(pv[0]%2)*RECT_TILE_W_DISP,(pv[0]//2)*RECT_TILE_H_DISP
            p2=(ox2+int(pv[1]*scale_rect), oy2+int(pv[2]*scale_rect))
            cv2.line(r,p2,p,(0,255,0),2)

    # fisheye polyline
    for i,(uf,vf) in enumerate(fish_pts):
        p=(int(uf/scale_fx), int(vf/scale_fy)) 
        cv2.circle(f,(p),4,(0,255,255),-1)
        if i:
            uf2,vf2=fish_pts[i-1]
            p2=(int(uf2/scale_fx),int(vf2/scale_fy))
            cv2.line(f,p2,p,(0,255,255),2)

    # BEV polyline
    for i,(bx,by) in enumerate(bev_pts):
        p=(int(bx*scale_BEV),int(by*scale_BEV))
        cv2.circle(b,p,4,(0,0,255),-1)
        if i:
            bx2,by2=bev_pts[i-1]
            p2=(int(bx2*scale_BEV),int(by2*scale_BEV))
            cv2.line(b,p2,p,(0,0,255),2)

    show(r,f,b)

canvas0 = np.ones((max(FISH_H_DISP, rect_grid_disp.shape[0]),
                   FISH_W_DISP + rect_grid_disp.shape[1],3),np.uint8)*255
canvas0[:FISH_H_DISP,:FISH_W_DISP]=fisheye_display
canvas0[:rect_grid_disp.shape[0],FISH_W_DISP:]=rect_grid_disp

cv2.namedWindow("Unified", cv2.WINDOW_NORMAL)
cv2.setMouseCallback("Unified", on_mouse)
cv2.imshow("Unified", canvas0); cv2.imshow("BEV", bev_display.copy())

while True:
    if cv2.waitKey(10)&0xFF==27: break
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# =========================
# PARAMETERS
# =========================
fisheye_params = {
    "width": 2592,
    "height": 1944,
    "cx": 1296.0,
    "cy": 972.0,
    "radius": 952.56,
    "theta_max_rad": 1.5708
}

cam_settings = [
    {"fov": 90.0, "pitch": 58.0, "yaw": -18.0},
    {"fov": 90.0, "pitch": 54.0, "yaw": 48.0},
    {"fov": 90.0, "pitch": 28.0, "yaw": 150.0},
    {"fov": 90.0, "pitch": 58.0, "yaw": 294.0},
]

# =========================
# INPUTS
# =========================
fisheye_img = cv2.imread("cam_4/cam_234_fisheye.png")
bev_229 = cv2.imread("229_fov/fov_165.jpg")      # MUST be 2560×2560
H_234_to_229 = np.load("Homography kafka_234_to_229_cam.npy")

assert bev_229.shape[:2] == (2560, 2560)

# =========================
# INTRINSICS (2560×2560)
# =========================
rect_w = rect_h = 2560
fov = 160.0
focal = (rect_w / 2) / np.tan(np.deg2rad(fov) / 2)

K_rect = np.array([
    [focal, 0, rect_w / 2],
    [0, focal, rect_h / 2],
    [0, 0, 1]
], np.float32)

f_fish = fisheye_params["radius"] / fisheye_params["theta_max_rad"]
K_fish = np.array([
    [f_fish, 0, fisheye_params["cx"]],
    [0, f_fish, fisheye_params["cy"]],
    [0, 0, 1]
], np.float32)

D_fish = np.zeros((4, 1), np.float32)

# =========================
# EXACT FUNCTION (UNCHANGED)
# =========================
def extract_directional_crop(
    fisheye_img, yaw_deg, pitch_deg,
    fov_deg=90, out_w=1280, out_h=720,
    fisheye_params=fisheye_params
):
    cx = fisheye_params["cx"]
    cy = fisheye_params["cy"]
    f_fish = fisheye_params["radius"] / fisheye_params["theta_max_rad"]

    yaw = np.deg2rad(yaw_deg)
    pitch = np.deg2rad(pitch_deg)
    fov = np.deg2rad(fov_deg)

    fx = (out_w / 2) / np.tan(fov / 2)
    fy = fx
    cx_o = out_w / 2
    cy_o = out_h / 2

    xv, yv = np.meshgrid(np.arange(out_w), np.arange(out_h))
    x_cam = (xv - cx_o) / fx
    y_cam = (yv - cy_o) / fy
    z_cam = np.ones_like(x_cam)

    norm = np.sqrt(x_cam**2 + y_cam**2 + z_cam**2)
    x_cam /= norm
    y_cam /= norm
    z_cam /= norm

    R_yaw = np.array([
        [np.cos(-yaw), -np.sin(-yaw), 0],
        [np.sin(-yaw),  np.cos(-yaw), 0],
        [0, 0, 1]
    ])

    R_pitch = np.array([
        [1, 0, 0],
        [0, np.cos(pitch), -np.sin(pitch)],
        [0, np.sin(pitch),  np.cos(pitch)]
    ])

    R_fix = np.array([
        [0, -1, 0],
        [1,  0, 0],
        [0,  0, 1]
    ])

    R = R_fix @ R_yaw @ R_pitch

    rays = np.stack((x_cam, y_cam, z_cam), axis=-1)
    rays = rays @ R.T

    theta = np.arccos(rays[..., 2])
    phi = np.arctan2(rays[..., 1], rays[..., 0])

    r = f_fish * theta
    u = cx + r * np.cos(phi)
    v = cy + r * np.sin(phi)

    return cv2.remap(
        fisheye_img,
        u.astype(np.float32),
        v.astype(np.float32),
        cv2.INTER_LINEAR
    )

# =========================
# RECT VIEWS (NO SCALING)
# =========================
views = [
    extract_directional_crop(
        fisheye_img,
        c["yaw"],
        c["pitch"]
    ) for c in cam_settings
]

rect_h, rect_w = views[0].shape[:2]

rect_grid = np.zeros((rect_h*2, rect_w*2, 3), np.uint8)
rect_grid[0:rect_h, 0:rect_w] = views[0]
rect_grid[0:rect_h, rect_w:] = views[1]
rect_grid[rect_h:, 0:rect_w] = views[2]
rect_grid[rect_h:, rect_w:] = views[3]

# =========================
# BEV DISPLAY SCALE ONLY
# =========================
scale_bev = 0.5
bev_disp = cv2.resize(bev_229, None, fx=scale_bev, fy=scale_bev)

# =========================
# CLICK HANDLER
# =========================
def on_mouse(event, x, y, *_):
    if event != cv2.EVENT_LBUTTONDOWN:
        return

    if x < rect_grid.shape[1]:
        return

    xr = x - rect_grid.shape[1]
    vid = (y >= rect_h) * 2 + (xr >= rect_w)

    xi = xr % rect_w
    yi = y % rect_h

    # rect point (UNSCALED)
    pt_rect = np.array([[[xi, yi]]], np.float32)

    und = cv2.fisheye.undistortPoints(
        pt_rect, K_fish, D_fish, P=K_rect
    )

    x229, y229 = cv2.perspectiveTransform(
        und, H_234_to_229
    )[0, 0]

    print("229 rect (2560 space):", x229, y229)

    disp = bev_disp.copy()
    cv2.circle(
        disp,
        (int(x229 * scale_bev), int(y229 * scale_bev)),
        6, (0, 0, 255), -1
    )
    cv2.imshow("229 BEV", disp)

# =========================
# UI
# =========================
canvas = np.zeros(
    (rect_grid.shape[0], rect_grid.shape[1]*2, 3),
    np.uint8
)

canvas[:, :rect_grid.shape[1]] = rect_grid
canvas[:, rect_grid.shape[1]:] = cv2.resize(
    fisheye_img,
    (rect_grid.shape[1], rect_grid.shape[0])
)

cv2.namedWindow("main", cv2.WINDOW_NORMAL)
cv2.setMouseCallback("main", on_mouse)

cv2.imshow("main", canvas)
cv2.imshow("229 BEV", bev_disp)

while cv2.waitKey(10) != 27:
    pass

cv2.destroyAllWindows()
```
